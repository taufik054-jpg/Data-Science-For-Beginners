# Advanced Notebook: CBR vs KNN

Perbandingan langsung antara:
- Case-Based Reasoning (CBR)
- K-Nearest Neighbors (KNN)

Tujuan:
Memahami perbedaan konsep, mekanisme, dan hasil.

## 🔹 Dataset

```python
import pandas as pd

data = {
    "demam":[1,1,1,0,0,1,0,1],
    "batuk":[1,1,0,1,0,1,0,0],
    "sakit_kepala":[0,1,1,0,1,1,0,1],
    "penyakit":["flu","flu","dbd","normal","normal","dbd","normal","flu"]
}

df = pd.DataFrame(data)
df
```

## 🔹 CBR Implementation

```python
# CBR similarity
def similarity(a,b):
    return sum([1 if x==y else 0 for x,y in zip(a,b)]) / len(a)

case_base = df.values.tolist()

new_case = [1,1,1]

results=[]
for row in case_base:
    sim = similarity(new_case, row[:3])
    results.append((row[3], sim))

sorted(results, key=lambda x:x[1], reverse=True)[:3]
```

## 🔹 KNN Implementation

```python
from sklearn.neighbors import KNeighborsClassifier

X = df[["demam","batuk","sakit_kepala"]]
y = df["penyakit"]

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X,y)

pred = model.predict([new_case])
pred
```

## 🔹 Probabilitas KNN

```python
model.predict_proba([new_case])
```

## 🔹 Perbandingan


| Aspek | CBR | KNN |
|------|-----|-----|
| Basis | Case | Data |
| Learning | Incremental | Training |
| Transparansi | Tinggi | Sedang |
| Adaptasi | Manual | Otomatis |
